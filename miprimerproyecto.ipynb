Ejercicio encontrar la nota final de cada materia, haciendo uso de finciones.

```python
def calcular_nota_materia(materia):
    c1 = (materia["notas"]["C1"]["lab1"] + materia["notas"]["C1"]["lab2"] + materia["notas"]["C1"]["lab3"]) / 3
    c2 = (materia["notas"]["C2"]["lab1"] + materia["notas"]["C2"]["lab2"] + materia["notas"]["C2"]["lab3"]) / 3
    c3 = (materia["notas"]["C3"]["lab1"] + materia["notas"]["C3"]["lab2"] + materia["notas"]["C3"]["lab3"]) / 3

    nota_final = (c1 + c2 + c3) / 3
    return nota_final


print(f"Estudiante: {persona['nombre']}\n" + "-" * 30)

for una_materia in persona["materias"]:
    resultado = calcular_nota_materia(una_materia)
    print("Materia:", una_materia["nombre"])
    print("Nota Final:", round(resultado, 2))
    print("-" * 30)
```

Calcular sobre el sueldo las deducciones por ley ISSS (3%), AFP(6.25%), ISR(Según tabla) e imprima el total deducciones y
sueldo neto. (20%)
Calcular el valor a pagar según el número de metros consumidos de agua potable (Ver tabla en la siguiente diapositiva).
(20%)
Desarrolle la tabla de meses acumulados a n meses. Si puede encontrar la fórmula matemática para evitar ciclos. (10%)

```python
sueldo = float(input("Ingrese el sueldo: "))

isss = sueldo * 0.03
afp = sueldo * 0.0625

baseGravable = sueldo - isss - afp

if baseGravable <= 550.00:
    isr = 0
elif baseGravable <= 895.24:
    isr = 17.67 + (baseGravable - 550.00) * 0.10
elif baseGravable <= 2038.10:
    isr = 60.00 + (baseGravable - 895.24) * 0.20
else:
    isr = 288.57 + (baseGravable - 2038.10) * 0.30

totalDeducciones = isss + afp + isr
sueldoNeto = sueldo - totalDeducciones

print(f"ISSS (3%): ${isss:.2f}")
print(f"AFP (6.25%): ${afp:.2f}")
print(f"ISR: ${isr:.2f}")
print(f"Total deducciones: ${totalDeducciones:.2f}")
print(f"Sueldo neto: ${sueldoNeto:.2f}")
```

```python
unidades = {
    "Longitud": {
        "base": "Metro",
        "factores": {
            "Metro": 1,
            "Kilometro": 1000,
            "Centimetro": 0.01,
            "Milimetro": 0.001,
            "Milla": 1609.34,
            "Yarda": 0.9144,
            "Pie": 0.3048,
            "Pulgada": 0.0254,
            "Vara (El Salvador)": 0.835464,
            "Legua": 5572.7,
        }
    },
    "Masa": {
        "base": "Kilogramo",
        "factores": {
            "Kilogramo": 1,
            "Gramo": 0.001,
            "Miligramo": 0.000001,
            "Tonelada metrica": 1000,
            "Libra": 0.453592,
            "Onza": 0.0283495,
            "Arroba": 11.3398,
            "Quintal": 45.3592,
            "Tonelada corta": 907.185,
            "Tonelada larga": 1016.05,
        }
    },
    "Almacenamiento": {
        "base": "Byte",
        "factores": {
            "Bit": 0.125,
            "Byte": 1,
            "Kilobyte": 1024,
            "Megabyte": 1024 ** 2,
            "Gigabyte": 1024 ** 3,
            "Terabyte": 1024 ** 4,
            "Petabyte": 1024 ** 5,
            "Kilobit": 128,
            "Megabit": 1024 ** 2 * 0.125,
            "Gigabit": 1024 ** 3 * 0.125,
        }
    },
    "Tiempo": {
        "base": "Segundo",
        "factores": {
            "Milisegundo": 0.001,
            "Segundo": 1,
            "Minuto": 60,
            "Hora": 3600,
            "Dia": 86400,
            "Semana": 604800,
            "Mes (30 dias)": 2592000,
            "Anio": 31536000,
            "Decada": 315360000,
            "Siglo": 3153600000,
        }
    },
    "Volumen": {
        "base": "Litro",
        "factores": {
            "Mililitro": 0.001,
            "Centilitro": 0.01,
            "Litro": 1,
            "Metro cubico": 1000,
            "Galon": 3.78541,
            "Cuarto": 0.946353,
            "Pinta": 0.473176,
            "Onza liquida": 0.0295735,
            "Barril": 158.987,
            "Taza": 0.24,
        }
    },
    "Area": {
        "base": "Metro cuadrado",
        "factores": {
            "Metro cuadrado": 1,
            "Decarea": 10,
            "Hectarea": 10000,
            "Kilometro cuadrado": 1000000,
            "Manzana (El Salvador)": 6988.96,
            "Tarea": 628.86,
            "Vara cuadrada": 0.698896,
            "Acre": 4046.86,
            "Pie cuadrado": 0.092903,
            "Yarda cuadrada": 0.836127,
        }
    },
}


def mostrar_menu_principal():
    print("\n===== CONVERSOR DE UNIDADES =====")
    for i, categoria in enumerate(unidades, start=1):
        print(f"{i}. {categoria}")
    print("0. Salir")


def mostrar_submenu(categoria):
    factores = unidades[categoria]["factores"]
    print(f"\n--- Unidades de {categoria} ---")
    for i, unidad in enumerate(factores, start=1):
        print(f"{i}. {unidad}")
    return list(factores.keys())


def convertir(categoria, unidad_origen, unidad_destino, valor):
    factores = unidades[categoria]["factores"]
    valor_en_base = valor * factores[unidad_origen]
    return valor_en_base / factores[unidad_destino]


def main():
    while True:
        mostrar_menu_principal()
        opcion = input("Seleccione una categoria: ")

        if opcion == "0":
            print("Saliendo...")
            break

        categorias = list(unidades.keys())
        try:
            indice = int(opcion) - 1
        except ValueError:
            print("Opcion invalida.")
            continue

        if indice < 0 or indice >= len(categorias):
            print("Opcion invalida.")
            continue

        categoria = categorias[indice]
        lista_unidades = mostrar_submenu(categoria)

        try:
            origen_idx = int(input("Unidad de origen (numero): ")) - 1
            destino_idx = int(input("Unidad de destino (numero): ")) - 1
            valor = float(input(f"Ingrese el valor en {lista_unidades[origen_idx]}: "))
        except (ValueError, IndexError):
            print("Entrada invalida, intente de nuevo.")
            continue

        resultado = convertir(categoria, lista_unidades[origen_idx], lista_unidades[destino_idx], valor)
        print(f"\n{valor} {lista_unidades[origen_idx]} = {resultado:.6f} {lista_unidades[destino_idx]}")


if __name__ == "__main__":
    main()
```

```python
print("=== Sistema de Triage ===")

inconsciente = input("¿El paciente esta inconsciente? (s/n): ").lower()
emergencia = input("¿Presenta una emergencia (trabajo de parto, deshidratacion severa, dificultad respiratoria severa)? (s/n): ").lower()
edad = int(input("Edad del paciente (en anios, 0 si es menor de 1 anio): "))

fiebre_deshidratacion = "n"
if edad < 5:
    fiebre_deshidratacion = input("¿Presenta fiebre, deshidratacion o dificultad para respirar? (s/n): ").lower()

embarazada = input("¿Es una mujer embarazada? (s/n): ").lower()
ent_descompensada = input("¿Tiene una enfermedad cronica (ENT) descompensada? (s/n): ").lower()
discapacidad = input("¿Tiene alguna discapacidad? (s/n): ").lower()
preventiva = input("¿La atencion que busca es preventiva (control, vacuna, etc.)? (s/n): ").lower()

if inconsciente == "s" or emergencia == "s" or (edad < 5 and fiebre_deshidratacion == "s"):
    prioridad = "I - ROJO"
    tiempo = "Atencion inmediata"
elif embarazada == "s" or edad < 1 or edad >= 60 or ent_descompensada == "s" or discapacidad == "s" or preventiva == "s":
    prioridad = "II - AMARILLO"
    tiempo = "Menos de 2 horas"
else:
    prioridad = "III - VERDE"
    tiempo = "2 a 4 horas"

print(f"\nPrioridad asignada: {prioridad}")
print(f"Tiempo de espera estimado: {tiempo}")
```

```python
# Función para convertir Fahrenheit a Celsius
def fahrenheit_a_celsius(fahrenheit):
    celsius = (fahrenheit - 32) * 5 / 9
    return celsius
fahrenheit = float(input("Ingrese la temperatura en grados Fahrenheit: "))

celsius = fahrenheit_a_celsius(fahrenheit)

print("La temperatura en Celsius es:", celsius, "°C")
```

```python
def convertirKelvin(c):
    return float(c) + 273.15


c = input("C: ")

print("Grados Kelvin:", convertirKelvin(c))
```

2. Celsius a Kelvin

```python
converKel = lambda c: float(c) + 273.15

c = input("C: ")

print("Grados Kelvin:", converKel(c))
```

3. Calcular el sueldo de un empleado

```python
sueldo = lambda h, vh: float(h) * float(vh) * 30

horas = input("Horas trabajadas por día: ")
valor = input("Valor por hora: ")

total = sueldo(horas, valor)

isss = total * 0.03
afp = total * 0.0725
isr = total * 0.10

neto = total - isss - afp - isr

print("Sueldo Bruto: $", total)
print("ISSS: $", isss)
print("AFP: $", afp)
print("ISR: $", isr)
print("Sueldo Neto: $", neto)
```

4. Elevar un número a su exponente

```python
potencia = lambda n, e: float(n) ** float(e)

numero = input("Número: ")
exponente = input("Exponente: ")

print("Resultado:", potencia(numero, exponente))
```

Ejercicio encontrar la nota final de cada materia, haciendo uso de finciones.

```python
def calcular_nota_materia(materia):
    c1 = (materia["notas"]["C1"]["lab1"] + materia["notas"]["C1"]["lab2"] + materia["notas"]["C1"]["lab3"]) / 3
    c2 = (materia["notas"]["C2"]["lab1"] + materia["notas"]["C2"]["lab2"] + materia["notas"]["C2"]["lab3"]) / 3
    c3 = (materia["notas"]["C3"]["lab1"] + materia["notas"]["C3"]["lab2"] + materia["notas"]["C3"]["lab3"]) / 3

    nota_final = (c1 + c2 + c3) / 3
    return nota_final


print(f"Estudiante: {persona['nombre']}\n" + "-" * 30)

for una_materia in persona["materias"]:
    resultado = calcular_nota_materia(una_materia)
    print("Materia:", una_materia["nombre"])
    print("Nota Final:", round(resultado, 2))
    print("-" * 30)
```

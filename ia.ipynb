```python
#libreriar a import
import tensorflow as tf
import pandas as pd
import numpy as np
```

```python
#leer el archivo
datos = pd.read_csv('/content/conversor de celcius.csv', sep=';', encoding='utf-8')
print(datos["Celcius"])
```

```python
#obtener los datso
c = datos["Celcius"]
f = datos["Fahrenheit"]
print(f)
```

```python
#Crear el modelo
modelo = tf.keras.Sequential()
modelo.add(tf.keras.layers.Dense(units=1, input_shape=[1]))
```

```python
#Compilar el modelo
modelo.compile(optimizer=tf.keras.optimizers.Adam(1.0), loss='mean_squared_error')
```

```python
#Entrear modelo
epocas = modelo.fit(c, f, epochs=200, verbose=1)
```

```python
resp = modelo.predict(np.array([10.0]))
print(resp)
```

1. Desarrolle un algoritmo en Python que calcule el IMPUESTO A LAS
ACTIVIDADES ECONOMICAS de acuerdo con la siguiente tabla. El monto
de la actividad económica debe ser introducido por el usuario. (30%)

```python
def calcular_impuesto(monto):
    tabla = [
        (0.01, 500.00, 1.5, 0),
        (500.01, 1000.00, 1.5, 3),
        (1000.01, 2000.00, 3, 3),
        (2000.01, 3000.00, 6, 3),
        (3000.01, 6000.00, 9, 2),
        (6000.01, 8000.00, 12, 2),
        (8000.01, 18000.00, 15, 2),
        (18000.01, 30000.00, 39, 2),
        (30000.01, 60000.00, 63, 1),
        (60000.01, 100000.00, 93, 0.8),
        (100000.01, 200000.00, 125, 0.7),
        (200000.01, 300000.00, 195, 0.6),
        (300000.01, 400000.00, 255, 0.45),
        (400000.01, 500000.00, 300, 0.4),
        (500000.01, 1000000.00, 340, 0.30),
        (1000000.01, 99999999.00, 490, 0.18)
    ]

    for desde, hasta, precio, adicional in tabla:
        if desde <= monto <= hasta:
            excedente = monto - desde
            impuesto = (excedente / 1000) * adicional + precio
            return impuesto

    return None
print("=== CÁLCULO DE IMPUESTO A LAS ACTIVIDADES ECONÓMICAS ===")
try:
    monto_usuario = float(input("Ingrese el monto de la actividad económica ($): "))
    if monto_usuario > 0:
        resultado = calcular_impuesto(monto_usuario)
        if resultado is not None:
            print(f"El valor a pagar por ${monto_usuario:,.2f} es: ${resultado:.2f}")
        else:
            print("El monto excede los límites de la tabla.")
    else:
        print("Por favor, ingrese un monto mayor a 0.")
except ValueError:
    print("[!] Error: Debe ingresar un valor numérico válido.")
```

2. Desarrolle el conversor de Área (Superficie) (El Salavador) haciendo uso de
los diccionarios y las funciones de usuario en Python que contenga las
siguientes opciones: (30%)
a. Pie Cuadrado
b. Vara Cuadrada
c. Yarda Cuadrada
d. Metro Cuadrado
e. Tareas
f. Manzana
g. Hectárea

```python
# Diccionario con las unidades de superficie de El Salvador y sus factores base en metros cuadrados
unidades = {
    1: ("Pie Cuadrado", 0.092903),
    2: ("Vara Cuadrada", 0.698742),
    3: ("Yarda Cuadrada", 0.836127),
    4: ("Metro Cuadrado", 1.0),
    5: ("Tareas", 436.7135),
    6: ("Manzana", 6987.416),
    7: ("Hectárea", 10000.0)
}

def mostrar_menu():
    print("\n--- MENÚ DE UNIDADES DE SUPERFICIE ---")
    for clave, datos in unidades.items():
        print(f"[{clave}] {datos[0]}")

def convertir():
    print("=== CONVERSOR DE ÁREAS (EL SALVADOR) ===")
    mostrar_menu()
    try:
        opcion_origen = int(input("\nElige el número de la unidad de ORIGEN: "))
        if opcion_origen not in unidades:
            print("[!] Opción de origen no válida.")
            return
        mostrar_menu()
        opcion_destino = int(input("\nElige el número de la unidad de DESTINO: "))
        if opcion_destino not in unidades:
            print("[!] Opción de destino no válida.")
            return

        cantidad = float(input(f"Ingresa la cantidad de [{unidades[opcion_origen][0]}]: "))

        if cantidad < 0:
            print("[!] La cantidad no puede ser negativa.")
            return
        valor_en_metros = cantidad * unidades[opcion_origen][1]
        resultado = valor_en_metros / unidades[opcion_destino][1]
        print("\n" + "="*40)
        print(f" [RESULTADO]: {cantidad} {unidades[opcion_origen][0]} ")
        print(f" Equivalen a: {resultado:,.4f} {unidades[opcion_destino][0]}")
        print("="*40)

    except ValueError:
        print("[!] Error: Debes ingresar números enteros para las opciones y valores numéricos válidos para la cantidad.")
if __name__ == "__main__":
    convertir()
```

3. Sistema estadístico, calcule: (20%)
a. Media aritmética
b. Mediana
c. Moda
d. Varianza
e. Desviación estándar
f. Desviación típica
g. Rango y
h. Frecuencia
Explicar con sus palabras la diferencia.

```python
class SistemaEstadistico:
    def __init__(self, datos):
        self.datos = sorted([float(x) for x in datos])
        self.n = len(self.datos)

    def media_aritmetica(self):
        return sum(self.datos) / self.n

    def mediana(self):
        medio = self.n // 2
        if self.n % 2 == 0:
            return (self.datos[medio - 1] + self.datos[medio]) / 2
        else:
            return self.datos[medio]

    def moda(self):
        frecuencias = self.frecuencia()
        max_frecuencia = max(frecuencias.values())
        modas = [k for k, v in frecuencias.items() if v == max_frecuencia]
        return modas

    def varianza(self):
        media = self.media_aritmetica()
        suma_cuadrados = sum((x - media) ** 2 for x in self.datos)
        return suma_cuadrados / self.n

    def desviacion_estandar(self):
        return self.varianza() ** 0.5

    def desviacion_tipica(self):
        return self.desviacion_estandar()

    def rango(self):
        return self.datos[-1] - self.datos[0]

    def frecuencia(self):
        conteo = {}
        for x in self.datos:
            val = int(x) if x.is_integer() else x
            conteo[val] = conteo.get(val, 0) + 1
        return conteo

print("--- CONFIGURACIÓN DEL SISTEMA ESTADÍSTICO ---")
entrada = input("Introduce los datos separados por coma (ej: 18,20,19,21) o presiona ENTER para usar los datos por defecto: ")

if entrada.strip() == "":
    muestra_datos = [18, 20, 19, 21, 20, 22, 19, 20, 25, 21]
    print("Usando datos por defecto...")
else:
    muestra_datos = [float(x.strip()) for x in entrada.split(",")]
est = SistemaEstadistico(muestra_datos)
activo = True
while activo:
    print("\n" + "=" * 40)
    print("      MENÚ DE OPCIONES ESTADÍSTICAS")
    print("=" * 40)
    print(f"Datos actuales: {est.datos}")
    print("1. Media aritmética")
    print("2. Mediana")
    print("3. Moda")
    print("4. Varianza")
    print("5. Desviación estándar")
    print("6. Desviación típica")
    print("7. Rango")
    print("8. Frecuencia")
    print("9. Ver todos los resultados de un solo golpe")
    print("10. Cambiar los datos de la serie")
    print("11. Salir")

    opcion = input("\nElige una opción (1-11): ")

    if opcion == '1':
        print(f"\n-> Media aritmética: {est.media_aritmetica():.2f}")
    elif opcion == '2':
        print(f"\n-> Mediana: {est.mediana()}")
    elif opcion == '3':
        print(f"\n-> Moda: {est.moda()}")
    elif opcion == '4':
        print(f"\n-> Varianza: {est.varianza():.2f}")
    elif opcion == '5':
        print(f"\n-> Desviación estándar: {est.desviacion_estandar():.2f}")
    elif opcion == '6':
        print(f"\n-> Desviación típica: {est.desviacion_tipica():.2f}")
    elif opcion == '7':
        print(f"\n-> Rango: {est.rango()}")
    elif opcion == '8':
        print(f"\n-> Frecuencias: {est.frecuencia()}")
    elif opcion == '9':
        print("\n--- TODOS LOS RESULTADOS ESTADÍSTICOS ---")
        print(f"Media aritmética: {est.media_aritmetica():.2f}")
        print(f"Mediana: {est.mediana()}")
        print(f"Moda: {est.moda()}")
        print(f"Varianza: {est.varianza():.2f}")
        print(f"Desviación estándar: {est.desviacion_estandar():.2f}")
        print(f"Desviación típica: {est.desviacion_tipica():.2f}")
        print(f"Rango: {est.rango()}")
        print(f"Frecuencias: {est.frecuencia()}")
    elif opcion == '10':
        nueva_entrada = input("Introduce los nuevos datos separados por coma: ")
        muestra_datos = [float(x.strip()) for x in nueva_entrada.split(",")]
        est = SistemaEstadistico(muestra_datos)
        print("¡Datos actualizados con éxito!")
    elif opcion == '11':
        print("\n¡Gracias por usar el sistema estadístico! Saliendo...")
        activo = False
    else:
        print("\n[!] Opción no válida. Por favor, elige un número del 1 al 11.")
```

Explicación de las diferencias:
Media aritmética: Es el promedio de los datos. Se suman todos los valores y el resultado se divide entre la cantidad total de datos.

Mediana: Es el valor que queda exactamente en el medio. Debes ordenar los números de menor a mayor y encontrar la posición

Moda: Es el número o valor que más se repite en todo el grupo. Puede haber más de una moda o ninguna.

Rango: Es la distancia entre el valor más alto y el más bajo. Se calcula restando el número menor del número mayor.

Varianza: Mide qué tan alejados están los datos respecto a la media. Se calcula elevando al cuadrado las diferencias de cada dato con la media. Su unidad de medida queda al cuadrado.

Desviación estándar (o desviación típica): Es exactamente lo mismo. Es la raíz cuadrada de la varianza. Te dice, en las mismas unidades originales, qué tan dispersos están los datos con respecto a la media.

Frecuencia: Es el número de veces que se repite un dato específico dentro de una lista o tabla.
